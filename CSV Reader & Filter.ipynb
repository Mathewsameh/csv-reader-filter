```python

# Csv Reader & Filter
# A program to read, add, delete, search, and filter student records using a CSV file.

# Display the Menu

print("=" * 20)

# Display option 1: Show all students.
print("1 - Show all student")

# Display option 2: Add a new student.
print("2 - Add student")

# Display option 3: Delete a student.
print("3 - Delete student")

# Display option 4: Filter students based on their GPA.
print("4 - Filter students by grades")

# Display option 5: Search for a student.
print("5 - Search for student")

# Display option 6: Exit the program.
print("6 - Exit")

# Print another separator line.
print("=" * 20)

# Ask the user to enter a menu option.
# strip() removes spaces from the beginning and end of the input.
theChoice = input("Please enter a number: ").strip()

print("\n")

# Repeat the loop if the input is not numeric,
# greater than 6, or less than 1.
while not theChoice.isnumeric() or int(theChoice) > 6 or int(theChoice) < 1:

    # Display an error message containing the invalid input.
    print(f"Error, {theChoice} not valid")

    print("=" * 20)

    # Ask the user to enter a valid number from 1 to 6.
    theChoice = input("Please enter a number correctly from 1 to 6 only: ").strip()

# Convert the user's choice from a string to an integer.
theChoice = int(theChoice)

# Execute this block if the user chooses option 1.
if theChoice == 1:

    # Open the CSV file in read mode.
    myFile = open("students.csv", "r")

    # Read the entire file and store its contents in a variable.
    showAllStudents = myFile.read()

    # Close the file after reading its contents.
    myFile.close()

    # Check whether the file is empty.
    if len(showAllStudents) == 0:

        # Display a message if there are no students.
        print("No students yet")

    else:

        # Display all the contents of the file.
        print(showAllStudents)
        

# Execute this block if the user chooses option 2.
elif theChoice == 2:

    # Open the CSV file in append mode.
    # Append mode adds new data without deleting existing records.
    myFile = open("students.csv", "a")

    # Ask for the student's name.
    # strip() removes extra spaces, and capitalize() formats the name.
    theNewName = input("Please, enter your name: ").strip().capitalize()

    # Ask for the student's age and convert the input to an integer.
    theNewAge = int(input("Please, enter your age: ").strip())

    # Ask for the student's GPA and convert the input to a float.
    theNewGpa = float(input("Please, enter your gpa: ").strip())

    # Ask for the student's country and format the input.
    theNewCountry = input("Please, enter your country: ").strip().capitalize()

    # Ask for the student's ID and convert the input to an integer.
    theNewId = int(input("Please, enter your id: ").strip())

    # Write the student's information to the CSV file.
    # An f-string inserts variable values into the text.
    # Commas separate the fields, and \n starts a new line.
    myFile.write(f"{theNewName}, {theNewAge}, {theNewGpa}, {theNewCountry}, {theNewId}\n")

    # Close the file after writing the student's information.
    myFile.close()

    # Display a success message containing the student's name.
    print(f"{theNewName} added successfully")

# Execute this block if the user chooses option 3.
elif theChoice == 3:

    # Open the CSV file in read mode.
    myFile = open("students.csv", "r")

    # Ask for the name of the student to delete.
    # Remove extra spaces and format the name.
    the_name_to_delete = input("Please, enter the name you want to delete: ").strip().capitalize()

    # Ask for the student's ID and convert it to an integer.
    the_id = int(input("Please, enter the id: ").strip())

    # Read all lines from the file and store them in a list.
    students = myFile.readlines()

    # Close the file after reading its contents.
    myFile.close()

    # Open the file in write mode.
    # Write mode clears the existing contents of the file.
    myFile = open("students.csv", "w")

    # Create a Boolean variable to track whether the student is found.
    # Initially, assume the student does not exist.
    studentFound = False

    # Loop through every student record stored in the list.
    for student in students:

        # Remove surrounding whitespace and split the record into fields.
        # Each field is separated by a comma followed by a space.
        data = student.strip().split(", ")

        # Check whether both the name and ID match the requested student.
        if data[0] == the_name_to_delete and int(data[4]) == the_id:

            # Mark the student as found.
            studentFound = True

            # Skip writing this record, effectively deleting the student.
            continue

        # Write the current record back to the file if it is not deleted.
        myFile.write(student)

    # Close the file after rewriting the remaining records.
    myFile.close()

    # Check whether a matching student was found.
    if studentFound:

        # Display a success message if the student was deleted.
        print("Student deleted successfully")

    else:

        # Display a message if no matching student was found.
        print("The student not found")

# Execute this block if the user chooses option 4.
elif theChoice == 4:

    # Open the CSV file in read mode.
    myFile = open("students.csv", "r")

    # Set the minimum GPA required to display a student.
    min_gpa = 2.5

    # Read the file one line at a time.
    for student in myFile:

        # Remove surrounding whitespace and split the record into fields.
        data = student.strip().split(", ")

        # Get the GPA from index 2 and convert it to a float.
        student_gpa = float(data[2])

        # Check whether the student's GPA is at least the minimum GPA.
        if student_gpa >= min_gpa:

            # Display the student's record if the condition is true.
            print(student.strip())

    # Close the file after processing all student records.
    myFile.close()

# Execute this block if the user chooses option 5.
elif theChoice == 5:

    # Open the CSV file in read mode.
    myFile = open("students.csv", "r")

    # Ask for the name of the student to search for.
    # Remove extra spaces and format the name.
    the_name_to_search = input("Please, enter the name you want to search: ").strip().capitalize()

    # Ask for the student's ID and convert it to an integer.
    the_id = int(input("Please, enter the id: ").strip())

    # Read all lines from the file and store them in a list.
    students = myFile.readlines()

    # Close the file after reading all records.
    myFile.close()

    # Create a Boolean variable to track whether the student is found.
    # Initially, assume the student does not exist.
    student_found = False

    # Loop through every student record in the list.
    for student in students:

        # Remove surrounding whitespace and split the record into fields.
        data = student.strip().split(", ")

        # Check whether both the student's name and ID match.
        if data[0] == the_name_to_search and int(data[4]) == the_id:

            # Display the matching student's record.
            print(student.strip())

            # Mark the student as found.
            student_found = True

    # Check whether no matching student was found.
    if not student_found:

        # Display a message if the search returns no results.
        print("Student not found")

# Execute this block if the user chooses option 6.
elif theChoice == 6:

    # Display a goodbye message before ending the program.
    print("Thank you, Goodbye")
```
